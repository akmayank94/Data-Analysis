##### Introduction to Functions
Definition:

A function is a block of code that performs a specific task.
Functions help in organizing code, reusing code, and improving readability.



```python
# defining a function

def function_name (parameters):
    """
    any comment explaining usabilty
    """
    # function body
    return expression
```

```python
# why do we need a fucntion
num = 24

if num % 2 == 0:
    print("even")
else:
    print("odd")
```

```python
# even / odd function
def even_or_odd(num):
    """
    This function finds even or odd
    """
    if num%2==0:
        print("even")
    else:
        print("odd")


# call the function
even_or_odd(4)
```

```python
# functions with multiple parameters
def add(a,b):
    return a+b 

result = add(3,4)
print(result)
```

```python
# default parameters

def greet(name="Guest"):
    return print(f"Hey {name}, Welcome to the Paradise!")

# without parameter giving
greet()

# with parameter 
greet("Mayank")
```

```python
# variable length arguements
# positional and keywords arguements

# instead of this
# def print_numbers(num1,num2,num3,num4,num5):

# do this
# args is best practice but you can write *yourname whatever you want

# positional arguements
def print_numbers(*args):
    for number in args:
        print(number)

```

```python
# this is positional arguements , means all the parameter given is referenced by *args
print_numbers(1,2,3,4,8,"mayank",True)
```

```python
# keywords arguement
# here all the arguements will be in key:value pair
def print_details(**kwargs):
    for key,value in kwargs.items():
        print(f"{key}:{value}")

print_details(name="Mayank",age="22",country = "India")
```

```python
def print_details (*args,**kwargs):
    for val in args:
        print(f"Positional arguement:{val}")

    for key,val in kwargs.items():
        print(f"{key}:{val}")

print_details(1,2,3,4,8,"mayank",True,name="Mayank",age="22",country = "India")
```

```python
# return statements
# we can return multiple parameters
def mult(a,b):
    return a*b,a,b

mult(6,4)
```
