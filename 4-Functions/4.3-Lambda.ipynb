##### Lambda Functions in Python
Lambda functions are small anonymous functions defined using the **lambda** keyword. They can have any number of arguments but only one expression. They are commonly used for short operations or as arguments to higher-order functions.



```python
# syntax
lambda arguments: expressions
```

```python

# def addition(a,b):
#     return a+b

# addition(3,4)
```

```python
# examples
# lambda function can be simply stored in variable like this , its anonymous

# addition lambda function
addition = lambda a,b:a+b
```

```python
print(type(addition))
```

```python
print(addition(2,9))
```

```python
# check even
check_even = lambda a:a%2==0

print(check_even(49))
```

```python
# addition one 3 numbers
 
add = lambda x,y,z: x+y+z

print(add(3,4,9))
```

```python
# without map()
# need to iterate to all element of list and square it

num = [1,2,3,4,5]

def squared(num):
    result = []
    for n in num:
        result.append((n,n**2))

    return result

squared(num)
```

```python
# map() -> applies a function to all the elements of the list

# map(function,iterables)
# map uses a lazy loading technique , it means its still now completely initialized

map(lambda x: x**2, num)
```

```python
list(map(lambda x: x**2, num))
```
