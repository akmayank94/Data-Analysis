#### Numpy
NumPy is a fundamental library for scientific computing in Python. It provides support for arrays and matrices, along with a collection of mathematical functions to operate on these data structures. In this lesson, we will cover the basics of NumPy, focusing on arrays and vectorized operations.

```python
# np is just a alias for numpy
import numpy as np

# creating array using numpy

# 1D
arr_1d = np.array([1,2,3])
print(f"1D array : {arr_1d}")
print(type(arr_1d))
print(arr_1d.shape) #-> single dimension thats why only one number available : 3,
```

```python
arr_2d = arr_1d
arr_2d.reshape(1,3)
print(arr_2d)
print(arr_2d.shape)
# -> this is called reshaping 1D into 2D 
# main differentiator is double brackets on opening and closing [[...],[...]] its a nested list
```

```python
# 2D array
arr_2d = np.array([[1,2,3,4],[5,6,7,8]])
print(arr_2d)
```

```python
# np.arrange(start,end,stepsize)
# start including , end excluding 
np.arange(0,11,2)
```

```python
# np.ones() -> need tuple as input 
# np.ones((rows,cols))
np.ones((3,4))
```

```python
# identity matrix  -> diagnal elements 1, remaining 0
# always square matrix soo single value as input
np.eye(5)
```

```python
# Attributes of numpy
arr = np.array([[1,2,3],[4,5,6]])

print("Array:\n",arr)
print("Shape: ",arr.shape)
print("Number of dimensions: ",arr.ndim)
print("Size (number of elements): ", arr.size)
print("Dtype: ",arr.dtype)
print("Item Size (in bytes): ",arr.itemsize)
```

```python
arr_mem = np.arange(10,dtype=np.int32)
arr_mem.dtype
```

```python
# astype -> change the datatype
arr_mem.astype(np.int64)
arr_mem.dtype
```

```python
# astype() does NOT modify the original array.
# Instead, it creates and returns a new array with the new datatype.
# thats why above didn't changed
# storing it maked change
arr_mem_64 = arr_mem.astype(np.int64)
arr_mem_64.dtype
```

```python
# Numpy Vectorize operations

arr1 = np.array([1,2,3,4,5])
arr2 = np.array([10,20,30,40,50])
```

```python
# element wise addition -> simply add arrays 
arr1+arr2
```

```python
# element wise subtraction
arr1-arr2
```

```python
arr2-arr1
```

```python
# element wise multiplication and division

print(arr1*arr2, "\n")
print(arr1/arr2)
```

```python
# universal function -> functions which applies to the entire array
arr = np.array([2,3,4,5,6])
```

```python
# squareroot
print(np.sqrt(arr))

# exponential
print(np.exp(arr))

# sine
print(np.sin(arr))

# natural log
print(np.log(arr))
```

```python
# some practical applications wrt arrays

# 1. statistical concept -> normalization
# make data to have mean=0 and standard deviation = 1

data = np.arange(1,6)
print(data)

# calculate mean and sd
mean = np.mean(data)
std = np.std(data)

# normalize the data
normalized_data = (data - mean)/std
print("Normalized data: ",normalized_data)
```

```python
# other
np.median(data)
```

```python
np.var(data)
```

```python
# Logical operations

data = np.arange(1,11)

data[(data>5) & (data<8)]
```
