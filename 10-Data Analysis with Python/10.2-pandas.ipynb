#### Pandas-DataFrame And Series
Pandas is a powerful data manipulation library in Python, widely used for data analysis and data cleaning. It provides two primary data structures: Series and DataFrame. A Series is a one-dimensional array-like object, while a DataFrame is a two-dimensional, size-mutable, and potentially heterogeneous tabular data structure with labeled axes (rows and columns).

```python
import pandas as pd
```

Series
- A Pandas Series is a one-dimensional array-like object that can hold any data type.
- It's similar to a column in a table.

```python
# regular way of creating series 
data = [1,2,3,4,5,'Smarty',3.14,True]
series = pd.Series(data)
print(type(series))
print()
print(series)
```

```python
# Creating Series from dictionary
# key as index
# value becomes series element

data = {
    'a':1,
    'b':2,
    'c':3
}
pd.Series(data)
```

```python
# giving custom index in form of list
data=[10,20,30]
index_for_data=['apple','banana','cherry']
pd.Series(data,index = index_for_data)
```

Dataframe
- DataFrame is a two-dimensional, size-mutable, and potentially heterogeneous tabular data structure 
with labeled axes (rows and columns).
- Looks like a table with rows and cols.

```python
# Creating Dataframe Using "dictionary of lists" 
data = {
    'name':['Mayank','Saloni','Tushar'],
    'age': [22,24,12],
    'city':['Delhi','Banglore','Florida']
}


df = pd.DataFrame(data)
df
```

```python
type(df)
```

```python
## Create a Data frame From a "List of Dictionaries"

data=[
    {'Name':'Mayank','Age':22,'City':'Bangalore'},
    {'Name':'Saloni','Age':24,'City':'Bangalore'},
    {'Name':'Bappy','Age':22,'City':'Bangalore'},
    {'Name':'Jack','Age':32,'City':'Bangalore'}
]

df=pd.DataFrame(data)
print(df)
print(type(df))
```

```python
data = {
    'Age': [25, 30, 22],
    'City': ['New York', 'London', 'Paris']
}

df = pd.DataFrame(data, index=['Alice', 'Bob', 'Charlie'])
print(df)
```

```python
df
```

```python
# loc -> label based with primary input as column/row names

# selecting single row
df.loc['Bob']
```

```python
# selecting single value
df.loc['Alice','City']
```

```python
print(df.loc['Bob','Age'])
```

```python
# slicing with loc -> inclusive
df.loc['Alice':'Bob','Age':'City']
```

```python
# iloc -> position based

# selecting single row
df.iloc[0]
```

```python
# selecting single value
df.iloc[1,1]
```

```python
# slicing in iloc -> Exclusive
df.iloc[0:1,1:2]
```
