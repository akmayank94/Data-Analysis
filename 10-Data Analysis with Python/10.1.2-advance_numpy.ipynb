##Numpy Arrays vs Python Lists

```python
# SPEED: Which is faster
# we will create 2 list and add each of its element and store in c list. Similarly, this will be done using np also and we will compare the time of completion
import time


a = [i for i in range(10000000)]
b = [i for i in range(10000000,20000000)]

c = []
start = time.time()
for i in range(len(a)):
  c.append(a[i]+b[i])
print(time.time()-start)
```

```python
import numpy as np

a = np.arange(10000000)
b = np.arange(10000000,20000000)

start = time.time()
c = a+b

print(time.time()-start)
```

numpy array won and its ~40 times faster than python lists

```python
# Memory
#lists
a = [i for i in range(10000000)]
import sys

sys.getsizeof(a)
```

```python
# np arrays
a = np.arange(10000000,dtype = np.int32)
sys.getsizeof(a)
```

honestly both take similar space but we can change the dtype of numpy as per the requirements

##Advanced Indexing

```python
a = np.arange(24).reshape(6,4)
a
```

```python
# Fancy Indexing -> we pass the list of number and those number represent index of rows
# this get majorly used in pandas
a[[0,2,3,5]]
```

```python
# another example , when we need selective columns
a[:,[0,2,3]]
```

```python
# Boolean Indexing -> when we need to extract using a logic (eg: extract all even numbers , its a logic not a selection of rows by user.)
# Its super important and very useful
import numpy as np
a = np.random.randint(1,100,24).reshape(6,4)
a

# NOTE: randint is mainly for extracting number in perticular range, like np.random.randint(range of start and end, how many number needed)
```

```python
# find all numbers greater than 50
a > 50
```

```python
# write way to filter out in boolean indexing
a[a>50]
```

```python
# find out even numbers
a[a % 2 == 0]
```

```python
# find all numbers which are greater than 50 and even in nature
a[(a % 2 == 0) & (a > 50)]
```

```python
# find all numbers which are NOT divisible by 7
a[~(a % 7 == 0)]
```
