##### Importing Modules in Python: Modules and Packages
In Python, modules and packages help organize and reuse code. Here's a comprehensive guide on how to import them.

```python
# importing a package
import math

math.sqrt(16)
```

```python
# importing functions from package so we can use that directly 
# eg: instead of writing math.sqrt() -> write sqrt() directly
from math import sqrt,pi

print(sqrt(25))
print(pi)
```

```python
# import numpy as np -> here, np is a standard alias for numpy package
import numpy as np

np.array([1,2,3])
```

```python
# importing every module or function
from math import *

sqrt(99)
```

```python
"""
Making custom package 
1. make a folder , lets say 'package'
2. make __init__.py file inside of it, its important to create while making a package
3. make any file and write functions you want to make for your package , lets say in maths.py I wrote add function
4. use it as from package.maths import add
"""

from package.maths import add
add(3,4)
```

```python
# other way
from package import maths

maths.add(22,8)
```

```python
from package.subpackage import mult

mult.multiplication(4,5)
```

```python
# error when not imported 
multiplication(22,4)
```

#### Conclusion
Importing modules and packages in Python allows you to organize your code, reuse functionalities, and keep your projects clean and manageable.
