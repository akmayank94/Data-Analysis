##### Standard Library Overview
Python's Standard Library is a vast collection of modules and packages that come bundled with Python, providing a wide range of functionalities out of the box. Here's an overview of some of the most commonly used modules and packages in the Python Standard Library.

```python
import array
arr = array.array('i',[1,2,3,4])
arr
```

```python
import math
print(math.sqrt(16))
print(math.pi)
```

```python
import random

# randint provide random value within the specified range
print(random.randint(1,10))
print(random.choice(['apple','banana','cherry']))
```

```python
# os -> File And Directory Access
import os

# getcurrentworkingdirectory
print(os.getcwd())
```

```python
os.mkdir('test_dir')
```

```python
# create new file
# os.O_CREAT: Create file if it doesn't exist
# os.O_WRONLY: Open for writing only
os.open("test_file.txt",flags=os.O_CREAT | os.O_WRONLY)
```

```python
# shutil -> High level operations on files and collection of files
import shutil
shutil.copyfile("source.txt","destination.txt")
```

```python
## Data Serialization -> converting your dictionary/dataframe etc into json
import json
data={'name':'Krish','age':25}

# json.dumps(input is dictionary)
json_str=json.dumps(data)
print(json_str)
print(type(json_str))

# json.loads(output is dumped list storing variable)
parsed_data=json.loads(json_str)
print(parsed_data)
print(type(parsed_data))
```

```python
## csv
import csv

# open example.csv as file object
with open('example.csv',mode='w',newline='') as file:
    writer=csv.writer(file)
    writer.writerow(['name','age'])
    writer.writerow(['Mayank',22])

with open('example.csv',mode='r') as file:
    reader=csv.reader(file)
    for row in reader:
        print(row)
```

```python
## datetime
from datetime import datetime,timedelta

now=datetime.now()
print(now)

yesterday=now-timedelta(days=1)

print(yesterday)
```

```python
## time
import time
print(time.time())
time.sleep(2)
print(time.time())
```

```python
## Regular expresiion
import re

pattern=r'\d+'
text='There are 123 apples 456'
match=re.search(pattern,text)
print(match.group())
```
