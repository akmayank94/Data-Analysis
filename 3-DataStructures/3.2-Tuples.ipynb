##### Introduction to Tuples
Explanation:

Tuples are ordered collections of items that are immutable.
They are similar to lists, but their immutability makes them different.


```python
# creating a tuple
empty_tuple = ()
print(empty_tuple)
print(type(empty_tuple))
```

```python
# using inbuild function for creating tuple
tpl = tuple()
print(tpl)
print(type(tpl))
```

```python
# list to tuple conversion
tpl = tuple([1,2,3,4,5])
print(tpl)

# tuple to list conversion
lst = list((1,2,3,4,5))
print(lst)
```

```python
# tuple also supports mixed type of element
mixed = (1,"helloworld",3.14,True)
print(mixed)
```

```python
# Accessing elements in tuple

mixed[0]
```

```python
# using loop
for el in mixed:
    print(el)
```

```python
mixed[::-1]
```

```python
# Tuple Operations

# concatenation
concatenation_tpl = tpl + mixed
print(concatenation_tpl)
```

```python
# this way only elements get multiplied 
# not the values of tuple without iterating is done
mixed * 2
```

```python
# Immutable nature of tuples
# Tuples are immutable, means elements can't be changed once assigned

lst = [1,2,3,4]
lst[0] = "Mayank"
print(lst)
```

```python
tpl[0] = "Saloni"
```

```python
# Tuple methods
print(tpl.count(3.14))

print(mixed.count(3.14))

print(tpl.index(3))
```

```python
# packing and unpacking tuple

# packing -> if multiple value are assigned to single variable with coma seperation, by default they get stored in form of tuple. Example:

packed = 1,"Hello",True
print(packed)
```

```python
# unpack the tuple
a,b,c = packed
print(a)
print(b)
print(c)
```

```python
# unpacking with *
numbers = (1,2,3,4,5,6)
first,*middle,last = numbers
print(first)
print(middle)
print(last)
```

```python
# Nested List
lst = [[1,2,3],["hello","mumu",3.4],[True,False,"BOMB"]]
print(lst[0])
print(lst[0][2])
print(lst[2][0:3])

# Nester Tuples
tpl = ((1,2,3),("Hello","Heaven"),(True,"c"))
print(tpl[0])
print(tpl[::])
print(tpl[2][0])
```

```python
# iterating over nester tuple
for sub_tuple in tpl:
    for item in sub_tuple:
        print(item, end = " ")
    print()
```

#### Conclusion
Tuples are versatile and useful in many real-world scenarios where an immutable and ordered collection of items is required. They are commonly used in data structures, function arguments and return values, and as dictionary keys. Understanding how to leverage tuples effectively can improve the efficiency and readability of your Python code.
