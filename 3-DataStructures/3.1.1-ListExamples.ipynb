#### Real-World Examples Using Lists in Python
Lists are one of the most commonly used data structures in Python, thanks to their versatility and ease of use. Here are several practical examples that illustrate their use in real-world scenarios

##### Example 1. Manage A To Do List
- Create a To Do List To Keep Track OF Tasks

```python
# creating a todo list
to_do_list = ["Cleaning the house","Go for a Walk","Pay the bills","Study for master's exam"]

# adding more tasks
to_do_list.append("Buying GTA-6 for my brother")
to_do_list.append("Calling my mom asking about her health")

# removing the completed tasks
to_do_list.remove("Cleaning the house")
to_do_list.remove("Go for a Walk")

# reminder for a task
if "Pay the bills" in to_do_list:
    print("Don't forget to pay before 11:59 PM")

# tasks which are still remaining 
for task in to_do_list:
    print(f"-{task}")
```

##### Example 2: Organizing Student Grades
- Create a list to store and calculate average grades for students

```python
grades = [92,44,78,72,80,10]

# adding new grades
grades.append(99)

print(f"Grade list: {grades}")

# calculating the avg grades 
avg_grade = sum(grades) / len(grades)
print(f"Average Grades: {avg_grade}")

# finding highest and lowest grades
highest = max(grades)
lowest = min(grades)

print(f"Highest Grade in the class: {highest}")
print(f"Lowest Grade in the class: {lowest}")
```

##### Example 3: Managing An Inventory
- Use a list to manage inventory items in a store

```python
# managing inventory
inventory = ["apple","mango","banana","grape","apple","apple","banana","kiwi","orange"]

# adding new items in inventory
inventory.append("watermelon")
inventory.append("watermelon")

# remove the item which is out of stock
inventory.remove("kiwi")

# checking if item is in stock
item = input("Which item do you want to check? (Enter 'EXIT' to exit the loop)")

if item in inventory:
    print(f"{item} in in stock. {inventory.count(item)} {item}s are present")
else:
    print(f"{item} is not in inventory. OUT OF STOCK!")

print("Inventory items which are available:")
inventory.sort()
for fruit in inventory:
    print(f"-{fruit}")
```
