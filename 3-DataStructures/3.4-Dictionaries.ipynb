##### Introduction to Dictionaries

Dictionaries are unordered collections of items. They store data in key-value pairs.
Keys must be unique and immutable (e.g., strings, numbers, or tuples), while values can be of any type.

```python
# creating dictionary
empty_dict = {}

print(type(empty_dict))
```

```python
# inbuild function
empty_dict = dict()
print(type(empty_dict))
```

```python
students = {
    "name":"Mayank",
    "age": 22,
    "grade":24
}

print(students)
print(type(students))
```

```python
# key should be unique
# single key is always used 
students = {
    "name":"Mayank",
    "name": 22,
    "grade":24
}

print(students)
```

```python
# accessing dict elements
students = {
    "name":"Mayank",
    "age": 22,
    "grade":'A'
}

print(students['grade'])
print(students['age'])
```

```python
# accessing using get() method
print(students.get('grade'))
print(students.get('last_name'))

# giving default sentence if not available
print(students.get('last_name',"Not Available"))
```

```python
# modify dict elements
# dict are mutable so you can update and delete.

students
```

```python
# updating value
students["age"] = 33

# added new key value
students["address"] = "New Delhi"
students
```

```python
# delete the key and value pair
del students["grade"]
```

```python
students
```

```python
# common dict methods

keys = students.keys()
print(keys)


values = students.values()
print(values)

items = students.items()
print(items)
```

```python
# shallow copy -> making copy but copy have different location so updating one will not affect other

students_copy = students
print(students)
print(students_copy)

# updating original
print("updating original only but it updates the copy too, problmatic.")
students["name"] = "Monkey"
print(students)
print(students_copy)
```

```python
# shallow copy example
students_copy1 = students.copy() # shallow copy
print(students)
print(students_copy1)

print("now changing original value")
students["name"] = "Saloni"
print(students)
print(students_copy1)
```

```python
# iterating over the dict

# using loops to iterate over dictionaries, values, keys, items

# iterating over keys
for key in students.keys():
    print(key)
```

```python
for value in students.values():
    print(value)
```

```python
# iterate over items
for key,value in students.items():
    print(key,value)
```

```python
# nested dictionaries

students = {
    "student1": {"name":"Mayank","age":22},
    "student2": {"name":"Saloni","age":22},
    "student3": {"name":"Mansi","age":24}
}

print(students)
```

```python
# access nested dictionary elements
print(students["student2"]["name"])
```

```python
# iterating over nested dictionary
for student_id,stundent_info in students.items():
    print(student_id,stundent_info)
    for key,value in stundent_info.items():
        print(key,value)
```

```python
# dictionary comprehension
square = {x:x**2 for x in range(1,6)}
square
```

```python
# conditional dict comprehension
even = {x:x**2 for x in range(1,11) if x%2==0}
print(even)
```

```python
# Practical examples

# use a dictionary to count the frequency of elements in list
numbers = [1,2,2,3,3,3,4,4,4,4]
frequency = {}

for number in numbers:
    if number in frequency:
        frequency[number] += 1
    else: 
        frequency[number] = 1

print(frequency)
```

```python
# merge two dictionaries into one

a1 = {"a":1,"b":2}
a2 = {"c":3,"d":4}
merged = {**a1,**a2}
print(merged)
```

#### Conclusion
Dictionaries are powerful tools in Python for managing key-value pairs. They are used in a variety of real-world scenarios, such as counting word frequency, grouping data, storing configuration settings, managing phonebooks, tracking inventory, and caching results. Understanding how to leverage dictionaries effectively can greatly enhance the efficiency and readability of your code.
