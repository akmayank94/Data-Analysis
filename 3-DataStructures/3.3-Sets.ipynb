#### Sets
Sets are a built-in data type in Python used to store collections of unique items. They are unordered, meaning that the elements do not follow a specific order, and they do not allow duplicate elements. Sets are useful for membership tests, eliminating duplicate entries, and performing mathematical set operations like union, intersection, difference, and symmetric difference.

```python
# creating a set
myset = {1,2,3,4,5,6}
print(myset)
print(type(myset))
```

```python
emptySet = set()
print(type(emptySet))
```

```python
# typecasting 
my_set = set([1,2,3,4,5])
print(my_set)
print(type(my_set))
```

```python
# duplicate storing
my_set = set([1,2,4,2,3,4,5])
print(my_set)
print(type(my_set))
```

```python
# Basic set operations

# adding elements (adding duplicate wont work)
my_set.add(7)
my_set

my_set.add(2)
my_set
```

```python
# removing elements using .remove()-> it throw an error if we try to remove the element which is not present
# discard wont give any error

my_set.remove(7)
my_set
```

```python
my_set.discard(7)
my_set
```

```python
my_set.remove(7)
my_set
```

```python
# pop method -> FIFO
removed_element = my_set.pop()
print(removed_element)
print(my_set)
```

```python
# clear
my_set.clear()
my_set
```

```python
# SET MEMBERSHIP TEST
my_set = {1,2,4,5,3}
print(3 in my_set)
print(7 in my_set)
```

```python
# Mathematical operations
s1 = {1,2,3,4,5,6}
s2 = {4,5,6,7,8,9}

# union
union_set = s1.union(s2)
print(union_set)
```

```python
# intersection
intersec = s1.intersection(s2)
print(intersec)
```

```python
# intersetion update -> find intersection and update in the s1
s1.intersection_update(s2)
print(f"s1 set after intersection update: {s1}")
```

```python
# difference
print(s1.difference(s2))
```

```python
s1
```

```python
s2.difference(s1)
```

```python
# symmetric diff -> all the unique elements are combined and common discarded
s1.symmetric_difference(s2)
```

```python
# Set methods

set1 = {1,2,3,4,5}
set2 = {3,4,5}
```

```python
# issubset -> is set1 a subset of set2? OR do all elemtn of set1 also belong to set2
set2.issubset(set1)
```

```python
# issuperset
set1.issuperset(set2)
```

```python
# counting unique words in text

text = "in this file, we are practicing sets"
words = text.split()

unique = set(words)
print(unique)
print(len(unique))
```

#### Conclusion
Sets are a powerful and flexible data type in Python that provide a way to store collections of unique elements. They support various operations such as union, intersection, difference, and symmetric difference, which are useful for mathematical computations. Understanding how to use sets and their associated methods can help you write more efficient and clean Python code, especially when dealing with unique collections and membership tests.
