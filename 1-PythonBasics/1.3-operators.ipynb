#### Arithematic Operators

```python
a = 10
b = 5

# addition and subtraction
add_result = a+b
subt_result = a-b

# multiplication and division
mul_result = a*b
div_result = a/b

# floor division -> round the result down to nearest whole number (towards -infinity side)
floor_div_result = a//b

# modulus division -> gives the remainder
modulus_result = a%b

# exponent 
exponent_result = a**b
```

```python
print(f'''
addition result = {add_result}
subtraction result = {subt_result}
multiplication result = {mul_result}
division result = {div_result}
floor division result = {floor_div_result}
modulus result = {modulus_result}
exponential result = {exponent_result}
''')
```

#### Logical Operators

```python
# And , Not , Or

x = True
y = True

and_op = x and y
or_op = x or y
not_op = not y

print(and_op, or_op, not_op)
```

### Simple Calculator

```python
## Simple Calculator

# Simple calculator
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

# Performing arithmetic operations
addition = num1 + num2
subtraction = num1 - num2
multiplication = num1 * num2
division = num1 / num2
floor_division = num1 // num2
modulus = num1 % num2
exponentiation = num1 ** num2

# Displaying results
print("Addition:", addition)
print("Subtraction:", subtraction)
print("Multiplication:", multiplication)
print("Division:", division)
print("Floor Division:", floor_division)
print("Modulus:", modulus)
print("Exponentiation:", exponentiation)
```
